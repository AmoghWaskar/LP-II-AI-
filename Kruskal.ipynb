```python
def find(parent,x):
  while parent[x]!=x:
    x=parent[x]
  return x
```

```python
def kruskal(edges,vertices):
  edges.sort(key=lambda x:x[2])

  mst=[]
  cost=0
  parent={v:v for v in vertices}

  for u,v,w in edges:
    pu=find(parent,u)
    pv=find(parent,v)

    if pu!=pv:
      parent[pu]=pv
      mst.append((u,v,w))
      cost=cost+w

  print("MST: ")
  for u,v,w in mst:
    print(u,"-",v,"=",w)

  print("Total Cost: ",cost)

```
