```python
def prim(graph,start):
  visited=[start]
  cost=0
  edges=[]

  while len(visited) < len(graph):

    min_edge=None
    min_cost=float('inf')

    for u in visited:
      for v,w in graph[u]:
        if v not in visited  and w<min_cost:
          min_cost=w
          min_edge=(u,v)

    u,v=min_edge
    visited.append(v)
    edges.append((u,v,min_cost))
    cost+=min_cost

  print("MST: ")
  for u,v,w in edges:
    print(f"{u}-{w}={v}")
  print("Total cost: ",cost)
```
