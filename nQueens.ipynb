```python
def solveNQueens(n):
  board=[-1]*n

  sols=[]

  col=[False]*n
  diag1=[False]*(2*n-1)
  diag2=[False]*(2*n-1)

  def isSafe(row,c):
    return not col[c] and not diag1[row+c] and not diag2[row-c+n-1]

  def solve(row):
    if row==n:
      sols.append(board.copy())

    for c in range(n):
      if isSafe(row,c):
        board[row]=c

        col[c]=True
        diag1[row+c]=True
        diag2[row-c+n-1]=True

        solve(row+1)

        col[c]=False
        diag1[row+c]=False
        diag2[row-c+n-1]=False

  solve(0)
  return sols
```

```python
def printBoard(soln):
  n=len(soln)

  for i in range(n):
    row=['.']*n
    row[soln[i]]='Q'
    print(" ".join(row))
  print()
```

```python
n=int(input("Enter value of N: "))
sols=solveNQueens(n)

print(f"Total Solutions: {len(sols)}")
for i,sol in enumerate(sols):
  print(f"Solution {i}: {sol}")
  printBoard(sol)
```
